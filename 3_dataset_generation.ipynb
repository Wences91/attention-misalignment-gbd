```python
import pandas as pd
```

```python
df_works = pd.read_csv('data/mental_disorders_works.tsv', sep='\t')
df_views = pd.read_csv('data/mental_disorders_wikiviews.tsv', sep='\t')
df_epi = pd.read_csv('data/IHME-GBD_2021_DATA-b26d6d1a-1.csv', sep=',')
```

```python
df_epi = df_epi.groupby(['cause_id', 'year'])['val'].sum().unstack(fill_value=0)
df_epi.columns = [f'val_{col}' for col in df_epi.columns]
df_epi['total'] = df_epi.sum(axis=1)
df_epi.reset_index(inplace=True)
df_epi
```

```python
df_works_views = df_works[['Cause ID', 'Cause Name', 'Level', 'works', 'works_2016', 'works_2017', 'works_2018', 'works_2019', 'works_2020', 'works_2021']].merge(df_views[['Cause ID', 'page_views', 'language', '2016', '2017', '2018', '2019', '2020', '2021']])
df_works_views
```

```python
df_works_views_epi = df_works_views.merge(df_epi, how='inner', left_on='Cause ID', right_on='cause_id')
df_works_views_epi
```

```python
df_works_views_epi.to_csv('data/mental_disorders_metrics.tsv', sep='\t', index=False)
```
